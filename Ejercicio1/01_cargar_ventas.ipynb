```python

# Identificar el catálogo actual de Databricks
catalogo = spark.sql("SELECT current_catalog()").first()[0]

print("Catálogo actual:", catalogo)
```

```python

# Nombre del esquema de trabajo
esquema = "ingenieria_datos_juanpa"

# Crear el esquema dentro del catálogo disponible
spark.sql(
    f"CREATE SCHEMA IF NOT EXISTS `{catalogo}`.`{esquema}`"
)

print(f"Esquema creado: {catalogo}.{esquema}")
```

```python

from pyspark.sql.types import (
    StructType, StructField,
    IntegerType, StringType, DateType, LongType
)
from datetime import date

# Datos ficticios de ventas
ventas = [
    (1, date(2026, 10, 1), "Laptop Lenovo", "Tecnología", 2, 2500000),
    (2, date(2026, 10, 2), "Mouse Logitech", "Tecnología", 5, 85000),
    (3, date(2026, 10, 3), "Teclado mecánico", "Tecnología", 3, 180000),
    (4, date(2026, 10, 4), "Monitor Samsung", "Tecnología", 2, 900000),
    (5, date(2026, 10, 5), "Silla de oficina", "Hogar", 2, 450000),
    (6, date(2026, 10, 6), "Escritorio", "Hogar", 1, 750000),
    (7, date(2026, 10, 7), "Lámpara LED", "Hogar", 4, 65000),
    (8, date(2026, 10, 8), "Cuaderno", "Papelería", 10, 12000),
    (9, date(2026, 10, 9), "Bolígrafo", "Papelería", 20, 3500),
    (10, date(2026, 10, 10), "Carpeta", "Papelería", 8, 8000),
    (11, date(2026, 10, 11), "USB 64GB", "Accesorios", 6, 45000),
    (12, date(2026, 10, 12), "Audífonos Bluetooth", "Accesorios", 3, 150000),
    (13, date(2026, 10, 13), "Cable HDMI", "Accesorios", 5, 25000),
    (14, date(2026, 10, 14), "Webcam HD", "Tecnología", 2, 220000),
    (15, date(2026, 10, 15), "Soporte de laptop", "Accesorios", 4, 70000),

    # Error 1: venta sin producto
    (16, date(2026, 10, 16), None, "Tecnología", 1, 100000),

    # Error 2: cantidad negativa
    (17, date(2026, 10, 17), "Mouse Logitech", "Tecnología", -3, 85000),

    # Error 3: duplicado idéntico de la venta 2
    (2, date(2026, 10, 2), "Mouse Logitech", "Tecnología", 5, 85000)
]

# Definir los tipos de datos
schema = StructType([
    StructField("id_venta", IntegerType(), False),
    StructField("fecha", DateType(), False),
    StructField("producto", StringType(), True),
    StructField("categoria", StringType(), True),
    StructField("cantidad", IntegerType(), False),
    StructField("precio_unitario", LongType(), False)
])

# Crear DataFrame
df_ventas = spark.createDataFrame(ventas, schema)

# Mostrar registros y estructura
display(df_ventas)
df_ventas.printSchema()

print("Total de registros:", df_ventas.count())
```

```python

# Nombre completo de la tabla Delta
tabla_bronze = "workspace.ingenieria_datos_juanpa.ventas_bronze"

# Guardar el DataFrame como tabla Delta
df_ventas.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(tabla_bronze)

print(f"Tabla creada correctamente: {tabla_bronze}")
```
